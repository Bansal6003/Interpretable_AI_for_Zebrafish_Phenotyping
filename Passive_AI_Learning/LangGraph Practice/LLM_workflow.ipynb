```python
#START
#llm_qa: question and answer and updates the state
#Shows langchain and langgraph work together

#Question: string
#Answer: String
```

```python
from langgraph.graph import StateGraph, START, END
from langchain_anthropic import ChatAnthropic
from typing import TypedDict
from dotenv import load_dotenv
import os

#os.environ["ANTHROPIC_API_KEY"] = ""

load_dotenv()
```

```python
model = ChatAnthropic(
    model="claude-haiku-4-5-20251001",
    temperature=0
)
```

```python
#Creat State
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    #extract question from state
    question = state['question']
    
    #form a prompt
    prompt = f"Answer the follwing question: {question}"
    
    #ask question to LLM
    #use .content only to get the answer and remove unnecessarty kwargs inside it
    answer = model.invoke(prompt).content
    
    #Update the answer in the State
    state['answer'] = answer
    
    return state
```

```python
#create Graph
graph = StateGraph(LLMState)

#add nodes
graph.add_node('llm_qa', llm_qa)

#add edges
graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)

#compile graph
workflow = graph.compile()
```

```python
#execute
initial_state = {'question': 'How far is moon from earth?'}

final_state = workflow.invoke(initial_state)

print(final_state)
print(final_state['answer'])
```
